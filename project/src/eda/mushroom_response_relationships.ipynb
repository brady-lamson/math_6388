# Overview

This notebook writes a ton of plots. Simple ones that compare all columns to the response to detect relationships.

```python
import polars as pl
import plotly.express as px

data_directory = "../../data/mushroom"
filename = "g_mushroom.parquet"
df = pl.read_parquet(f"{data_directory}/{filename}")
```

```python
df.glimpse()
```

```python
boolean_columns = df.select(pl.col(pl.Boolean)).columns
write = False
for column in df.columns:
    if column == "poisonous":
        continue

    elif column in boolean_columns:
        fig = px.density_heatmap(
            df,
            x="poisonous",
            y=column,
            text_auto=True,
            color_continuous_scale="Greys",
            title=f"Response breakdown for {column}"
        )

    else:
        fig = px.histogram(
            df,
            x=column,
            color="poisonous",
            barmode="group",
            title=f"Response breakdown for {column}"
        )

    if write:
        fig.write_image(f"../../images/mushroom/{column}_vs_response.png")
    fig.show()
```
