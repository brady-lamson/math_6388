# Context and Setup

```python
import torch
import torchvision
from jupyter_builder.jupyterlab_semver import test_set
from torchvision.transforms import v2

# ImageNet standard normalization and resizing
transform = v2.Compose([
    v2.Resize((224, 224)), # Resize CIFAR10 images to fit the pretrained model
    v2.ToImage(),
    v2.ToDtype(torch.float32, scale=True),
    v2.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

batch_size = 32

print("Loading Training Set")
trainset = torchvision.datasets.CIFAR10(root='./data', train=True,
                                        download=True, transform=transform)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=batch_size,
                                          shuffle=True, num_workers=8)

print("Loading Testing Set")
testset = torchvision.datasets.CIFAR10(root='./data', train=False,
                                       download=True, transform=transform)
testloader = torch.utils.data.DataLoader(testset, batch_size=batch_size,
                                         shuffle=False, num_workers=8)

classes = ('plane', 'car', 'bird', 'cat',
           'deer', 'dog', 'frog', 'horse', 'ship', 'truck')
```

```python
import torch.nn as nn
import torch.optim as optim
from torchvision.models import resnet18, ResNet18_Weights

# 1. Load the pre-trained ResNet18 model
weights = ResNet18_Weights.DEFAULT
net = resnet18(weights=weights)

# 2. Freeze the base layers so we don't destroy the pre-trained features
for param in net.parameters():
    param.requires_grad = False

# 3. Replace the final fully connected layer (fc) to output 10 classes instead of 1000
# By default, new layers have requires_grad=True
num_ftrs = net.fc.in_features
net.fc = nn.Linear(num_ftrs, 10)

# Move model to GPU if available (Pretrained models are larger, GPU helps!)
device = torch.device(
    "cuda" if torch.cuda.is_available()
    else "mps" if torch.backends.mps.is_available()
    else "cpu"
)
net = net.to(device)

# Define loss and optimizer (only optimizing the new fc layer)
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(net.fc.parameters(), lr=0.001)
```

```python
print(f"Device: {device}")
print(f"Batch size: {trainloader.batch_size}")
print(f"Batches per epoch: {len(trainloader)}")
print(f"Dataset size: {len(trainloader.dataset)}")
```

```python
import os.path
PATH = './cifar_transfer_net.pt'

if os.path.isfile(PATH):
    print(f"Model found at <{PATH}>, loading model...")
    net.load_state_dict(torch.load(PATH, weights_only=True))
else:
    print(f'Model not found at <{PATH}>, training on {device}...')

    for epoch in range(2):  # loop over the dataset multiple times

        running_loss = 0.0
        for i, data in enumerate(trainloader, 0):
            # get the inputs; data is a list of [inputs, labels]
            inputs, labels = data[0].to(device), data[1].to(device)

            # zero the parameter gradients
            optimizer.zero_grad()

            # forward + backward + optimize
            outputs = net(inputs)
            loss = criterion(outputs, labels)

            loss.backward()
            optimizer.step()

            # print statistics
            running_loss += loss.item()
            if i % 200 == 199:    # print every 200 mini-batches
                print(f'[{epoch + 1}, {i + 1:5d}] loss: {running_loss / 200:.3f}')
                running_loss = 0.0

    print('Finished Training, saving model')
    torch.save(net.state_dict(), PATH)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Helper to un-normalize and show images for testing
def imshow(img):
    img = img * torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1) + torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
    npimg = img.numpy()
    return plt.imshow(np.transpose(npimg, (1, 2, 0)))

# Get a batch of test data
dataiter = iter(testloader)
images, labels = next(dataiter)

# print images
fig = imshow(torchvision.utils.make_grid(images[:4]))
fig.show()
print('GroundTruth: ', ' '.join(f'{classes[labels[j]]:5s}' for j in range(4)))

net.eval() # Set to evaluation mode

# Predict
images = images.to(device)
outputs = net(images)
_, predicted = torch.max(outputs, 1)

print('Predicted:   ', ' '.join(f'{classes[predicted[j]]:5s}' for j in range(4)))
```

```python
correct = 0
total = 0

# Set network to evaluation mode and disable gradient calculation
net.eval()
with torch.no_grad():
    for data in testloader:
        images, labels = data[0].to(device), data[1].to(device)

        # calculate outputs by running images through the network
        outputs = net(images)

        # the class with the highest energy is what we choose as prediction
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

print(f'Accuracy of the pretrained network on the 10000 test images: {100 * correct // total} %')
```

## 1. **Compute predictive entropy for every test image.**
   - Use `net.eval()` and `torch.no_grad()`, as in the previous cell.
   - Apply softmax to the output logits to obtain class probabilities.
   - For each image $\mathbf{x}_i$, calculate
     $$
     H_i=-\sum_{k=1}^{K}
     p_\theta(k\mid\mathbf{x}_i)\log p_\theta(k\mid\mathbf{x}_i).
     $$
   - Store the entropy, image, predicted label, and true label together so they remain correctly matched.
   - Ensure your computation handles probabilities near zero without producing numerical errors.

```python
import polars as pl

# Goal here is to save each batch of results as a polars dataframe, then union batches together iteratively
# I track batch and within batch ids for my own diagnostic purposes
def collect_batch_results(batch_index, outputs, labels, batch_size: int):
    probabilities = torch.softmax(outputs, dim=1)
    _, predicted = torch.max(probabilities, 1)
    # torch.special.xlogy calculates plogp - https://docs.pytorch.org/docs/2.14/special.html#torch.special.xlogy
    # It also controls for the case when the input is 0.
    batch_plogp = torch.special.xlogy(probabilities, probabilities)
    batch_entropy = -1 * torch.sum(batch_plogp, dim=1)
    batch_index = torch.full((batch_size,), batch_index)
    within_batch_index = torch.arange(0, batch_size)

    result_dict = {
        "batch_index": batch_index.tolist(),
        "within_batch_index": within_batch_index.tolist(),
        "predicted_label": predicted.tolist(),
        "true_label": labels.tolist(),
        "entropy": batch_entropy.tolist(),
    }
    return pl.from_dict(result_dict)

df = pl.DataFrame()
# Set network to evaluation mode and disable gradient calculation
net.eval()
with torch.no_grad():
    for i, data in enumerate(testloader):
        if i % 25 == 0:
            print(f"Beginning iteration {i} ---")

        images, labels = data[0].to(device), data[1].to(device)
        batch_size = len(labels)

        # calculate outputs by running images through the network
        outputs = net(images)
        batch_df = collect_batch_results(batch_index=i, outputs=outputs, labels=labels, batch_size=batch_size)

        if df.is_empty():
            df = batch_df
        else:
            df = pl.union([df, batch_df])

df = df.with_row_index("id")
```

```python
# Order by entropy in descending order, pull the top 20 rows, map the label numbers back to their strings
class_map = dict(enumerate(classes))
df_entropy = (
    df
    .sort(pl.col("entropy"), descending=True)
    .with_columns(
        pl.col("predicted_label").replace_strict(class_map).alias("predicted_label"),
        pl.col("true_label").replace_strict(class_map).alias("true_label")
    )
    .head(20)
)
df_entropy.glimpse()
```

```python
# Shamelessly copy paste previous helper, literally for convenience so I don't have to keep scrolling up to look at it.
# NOTE: Adjusted to return the figure, so it can be modified before fig.show()
def imshow(img):
    img = img * torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1) + torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
    npimg = img.numpy()
    return plt.imshow(np.transpose(npimg, (1, 2, 0)))

# Get images and create the image grid ---
# This is a hacky way to do this to be honest, but it's just homework...
images_top20 = []
ids = df_entropy.get_column("id").to_list()
for id in ids:
    images_top20.append(testset[id][0])

nrow = 4
ncol = 5
grid = torchvision.utils.make_grid(images_top20, nrow=nrow+1)

plt.figure(figsize=(15, 12))
imshow(grid)

# Create the text captions for each image ---
for index, df_row in enumerate(df_entropy.iter_rows(named=True)):
    # Below code dictates where the image captions will land
    # It basically points to the correct part of the grid
    plot_row, plot_column = divmod(index, 5)

    plt.text(
        (plot_column + 0.5) / ncol,    # Center of column
        1 - (plot_row + 0.1) / nrow,   # Top of the row
        f'True: {df_row["true_label"]} | Pred: {df_row["predicted_label"]}\n'
        f'Correct: {df_row["true_label"] == df_row["predicted_label"]}\n'
        f'Entropy: {df_row["entropy"]:.3f}',
        transform=plt.gca().transAxes, # Allows plot to interpret text coordinates correctly
        ha="center",
        va="center",
        color="white",
        fontsize=8,
        bbox=dict(facecolor="black", alpha=0.7)
    )

plt.axis("off")
plt.show()
```
